```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
df = spark.read.table("publicholidays")
display(df)
```

```python
display(df.filter("isPaidTimeoff is null").limit(20) )
```

```python
display(df.filter("isPaidTimeOff IS NOT NULL").limit(20))
```

```python
from pyspark.sql.functions import col, to_date, year, month

df_clean = (
    df.dropDuplicates()
      .withColumn("holidayDate", to_date(col("date")))
      .withColumn("year", year(col("holidayDate")))
      .withColumn("month", month(col("holidayDate")))
      .drop("date")
)
display(df_clean)
```

```python
df_clean.write.format("delta").mode("overwrite").saveAsTable("Second_Lakehouse.dbo.holidays_silver")
```

```python
df_silver = spark.read.table("Second_Lakehouse.dbo.holidays_silver")
display(df_silver)


df_gold = df_silver.groupBy("countryOrRegion", "year").count().withColumnRenamed("count", "holidayCount")

display(df_gold)
```

```python
df_gold.write.format("delta").mode("overwrite").saveAsTable("Second_Lakehouse.dbo.holidays_by_country_year")
```
